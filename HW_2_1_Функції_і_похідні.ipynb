1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
from matplotlib import pyplot as plt
import numpy as np
import sympy as sp

x_sym = sp.symbols('x')
f_expr = 10 * x_sym - 5
derivative_expr = sp.diff(f_expr, x_sym)

print(f'Function: f(x) = {f_expr}')
print(f'Derivative: f\'(x) = {derivative_expr}')

x = np.linspace(0, 10, 100)
f_lin = 10 * x - 5
derivative_val = float(derivative_expr)

plt.figure(figsize=(10, 5))
plt.plot(x, f_lin, label='f(x) = 10x - 5', color='blue')
plt.axhline(y=derivative_val, label=f"f'(x) = {derivative_val:.0f}", color='green')
plt.xlabel('x (hours)')
plt.ylabel('y (score)')
plt.title('Function and its derivative')
plt.legend()
plt.grid(True)
plt.show()

print('Висновок: Оцінка зростає рівномірно: кожна додаткова година навчання збільшує оцінку на 10 балів (це і є похідна  f′(x)=10')
```

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
x_sym2 = sp.symbols('x')
f_expr2 = sp.sin(x_sym2)
derivative_expr2 = sp.diff(f_expr2, x_sym2)

x_sym3 = sp.symbols('x')
f_expr3 = sp.ln(x_sym3)
derivative_expr3 = sp.diff(f_expr3, x_sym3)

def plot_function_and_derivative(x_vals, x_symbol, func_expr, deriv_expr, title):
    f_num = sp.lambdify(x_symbol, func_expr, 'numpy')
    d_num = sp.lambdify(x_symbol, deriv_expr, 'numpy')

    plt.figure(figsize=(10, 5))
    plt.plot(x_vals, f_num(x_vals), label=f'f(x) = {sp.sstr(func_expr)}', color='blue')
    plt.plot(x_vals, d_num(x_vals), label=f"f'(x) = {sp.sstr(deriv_expr)}", color='green')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title(title)
    plt.legend()
    plt.grid(True)
    plt.show()

plot_function_and_derivative(x, x_sym2, f_expr2, derivative_expr2, 'Sin Function and its Derivative')
plot_function_and_derivative(x, x_sym3, f_expr3, derivative_expr3, 'Logarithmic Function and its Derivative')

print('Висновок:')
print('sin(x): похідна cos(x) випереджає sin(x) на π/2. Там, де sin зростає — cos додатній;')
print('ln(x): похідна 1/x завжди додатня (функція зростає), але зменшується — кожна наступна одиниця x дає все менший приріст.')
```


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
x_sym4 = sp.symbols('x')
f_expr4 = x_sym4 ** 2 - 5 * x_sym4 + 1
derivative_expr4 = sp.diff(f_expr4, x_sym4)

x4 = np.linspace(0, 100, 100)
plot_function_and_derivative(x4, x_sym4, f_expr4, derivative_expr4, 'Quadratic Function and its Derivative')

print('Висновок: Витрати зростають квадратично. Похідна f\'(x) = 2x - 5 — лінійна: при x < 2.5 витрати ще зменшуються, при x > 2.5 — зростають все швидше зі збільшенням виробництва.')
```
